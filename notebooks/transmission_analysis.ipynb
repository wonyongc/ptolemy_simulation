# Transmission Analysis Example
Uses `ptolemy_simulation.analysis.transmission` outputs for theta/phi sweeps.

```python
from pathlib import Path
from ptolemy_simulation.analysis.transmission import scan_theta_phi_files, aggregate_theta, plot_aggregate

input_dir = Path('generated/analysis/transmission_inputs')
mapping = scan_theta_phi_files(input_dir)
theta_data = aggregate_theta(mapping, particles_per_run=10200)
plot_aggregate(theta_data, Path('generated/analysis/transmission_notebook.png'))
print('theta groups:', len(theta_data))
```
